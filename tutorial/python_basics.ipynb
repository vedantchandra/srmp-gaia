### Lists and loops

A list holds several values in order. Python counts from **0**, so `planets[0]` is the first item.

```python
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune"]
print(len(planets))     # how many items
print(planets[0])       # first item
print(planets[-1])      # last item
print(planets[0:4])     # items 0, 1, 2, 3 (the end number is not included)
```

A `for` loop runs the same code for each item in a list. The indented lines are inside the loop.
Python uses the indentation to know what's inside, so it matters.

`zip` lets you loop over two lists side by side.

```python
distance_from_sun_au = [0.39, 0.72, 1.00, 1.52, 5.20, 9.54, 19.2, 30.1]

for name, a in zip(planets, distance_from_sun_au):
    print(name, "is", a, "AU from the Sun")
print("done")   # not indented, so this runs once, after the loop
```

```python
rng = np.random.default_rng(seed=42)    # the seed makes the "random" numbers repeatable

# pretend we measured a star's parallax 10 times: true value 10 mas, error 0.5 mas
measurements = rng.normal(loc=10.0, scale=0.5, size=10)
print(measurements)
print("mean:", np.mean(measurements))
print("median:", np.median(measurements))            # the middle value
print("standard deviation:", np.std(measurements))   # how spread out they are
```

## 5. scipy: fitting a model to data

Now something close to the real project.

A star with an unseen planet wobbles back and forth, on top of its steady drift across the sky.
We'll simulate this in one direction. The star's position at time $t$ is

$$ x(t) = x_0 + \mu\, t + A \sin\!\left(\frac{2\pi t}{P} + \phi\right) $$

$x_0$ is where it starts, $\mu$ is how fast it drifts, and $A$, $P$ and $\phi$ are the size, period
and phase of the wobble.

```python
def wobble_model(t, x0, mu, A, P, phi):
    """Position [mas] of a drifting, wobbling star at time t [years]."""
    return x0 + mu * t + A * np.sin(2 * np.pi * t / P + phi)

# the "true" values, which we'll pretend not to know
x0_true, mu_true, A_true, P_true, phi_true = 0.0, 5.0, 0.5, 2.3, 1.0

# Gaia sees each star about 70 times over 5 years, at irregular times
rng = np.random.default_rng(1)
t_obs = np.sort(rng.uniform(0, 5, size=70))
error = 0.1                                     # measurement error, mas
x_obs = wobble_model(t_obs, x0_true, mu_true, A_true, P_true, phi_true) + rng.normal(0, error, size=70)

fig, ax = plt.subplots()
ax.errorbar(t_obs, x_obs, yerr=error, fmt="o")
ax.set_xlabel("Time [years]")
ax.set_ylabel("Position [mas]")
plt.show()
```

It looks like a straight line, because the drift is much bigger than the wobble. Let's fit a
straight line first, and look at what's left over (the **residuals**).

`curve_fit` takes a model function and the data, and finds the best values of the model's
numbers. It returns two things. The first, `popt`, holds the best-fit values in the same order as
the function's inputs. (Ignore the second one for now.)

```python
from scipy.optimize import curve_fit

def line(t, x0, mu):
    return x0 + mu * t

popt, _ = curve_fit(line, t_obs, x_obs)
print("best-fit x0 and mu:", popt)

residuals = x_obs - line(t_obs, popt[0], popt[1])

fig, ax = plt.subplots()
ax.errorbar(t_obs, residuals, yerr=error, fmt="o")
ax.axhline(0, color="k")
ax.set_xlabel("Time [years]")
ax.set_ylabel("Data minus line [mas]")
plt.show()
```

The residuals aren't random: there's a wave. A straight line isn't enough.

## 6. astropy: units (optional, if you have time)

Mixing up units is one of the most common bugs in astronomy. `astropy.units` attaches units to
numbers and converts them for you with `.to()`.

```python
import astropy.units as u

d = 1.30 * u.pc
print(d.to(u.lyr))
print(d.to(u.km))
print((1 * u.M_jup).to(u.M_earth))       # Jupiter's mass in Earth masses
```

### How big is a planet's wobble?

A planet and its star both orbit their shared center of mass. The star's orbit is smaller than the
planet's by the ratio of their masses:

$$ a_\star = a \times \frac{m_\text{planet}}{M_\text{star}} $$

**Try it.** Jupiter is 5.2 AU from the Sun. How big is the Sun's orbit because of Jupiter? Then,

Gaia measures bright stars to about 0.05 mas each time it looks, and it looks dozens of times, so a
Jupiter like this is easy for Gaia to see.
